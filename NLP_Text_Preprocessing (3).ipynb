# 1. Stop Words Removal

Stop words in NLP are common words like "the," "is," "and," and "in" that are filtered out during text processing because they carry little semantic value or are otherwise insignificant for specific tasks.

**Common English Examples:**

* Pronouns: I, you, he, she, it, they
* Prepositions: in, on, at, by, with
* Conjunctions: and, but, or, so

Why do we remove them?

The primary goal of removing stop words is to focus the "attention" of a model on the words that actually define the topic.

1. Reduces Noise: Filters out generic words so models can focus on meaningful terms.
2. Boosts Efficiency: Shrinks dataset size, making processing and training significantly faster.

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
```

```python
stop_words = stopwords.words('english')
print(f"Total English stop words in NLTK: {len(set(stop_words))}\n")
```

```python
print(f"First 10 stop words: {list(stop_words)[:10]}")
```

### Example: Removing Stop Words from a Sentence

```python
# Sample text
text = "This is an example sentence demonstrating the removal of stop words in natural language processing"

# Tokenize the text (break it into individual words)
# We make it lowercase first so it matches the lowercase stopwords list perfectly
words = word_tokenize(text.lower())
words
```

```python
# Filter out the stop words
filtered_sentence = []
for w in words:
    if w not in stop_words:
        filtered_sentence.append(w)

# Original Sentence
print(f"Before stop word -> {text}")

# After removing stop words
print(f"After stop word -> {" ".join(filtered_sentence)}")
```

# **2. Stemming**

Stemming is a text preprocessing technique in NLP that reduces inflected or derived words to their root form, known as a `stem`, by removing prefixes and suffixes. One of the most popular methods was developed by `Martin Porter`, known as the **Porter Stemmer**

## **How it Works**
Stemming uses a set of fixed rules to strip characters from the end of a word. It doesn't actually "understand" the language; it just follows instructions like "If the word ends in 'ing', remove it."
### Examples:
- playing -> play
- played -> play
- plays -> play
- studies -> studi

## **Advantages and Limitations**
Stemming is fast and useful, but it’s considered a rough or imprecise technique because it relies on fixed rules instead of true language understanding.

* **Over-stemming:** When two different words are reduced to the same root incorrectly.
    * *Example:* "University" and "Universe" might both be reduced to **"Univers"**, even though they mean different things.
* **Under-stemming:** When two words that should be the same are left separate.
    * *Example:* "analyze" and "analysis" might remain different because the rules aren't sophisticated enough to link them.

```python
from nltk.stem import PorterStemmer

text = "The leaders handled the obvious problems in their surprising situation."
filtered_text = [word for word in word_tokenize(text.lower()) if word not in stop_words] # clear text using stop word
filtered_text
```

```python
porter = PorterStemmer()

print(f"{'Original Word':<15} | {'Porter Stem':<15}")
print("-" * 30)

for word in filtered_text:
    if not word.isalpha(): # Skip punctuation
        continue

    porter_stem = porter.stem(word)

    print(f"{word:<15} | {porter_stem:<15}")
```

# **3. Lemmatization**



“Imagine you’re searching for ‘run’, but the text contains ‘running’, ‘ran’, and ‘runs’.  
Would you miss them just because they look different?”

💡 Problem:  
Computers treat them as different words, even though they have the same meaning.

### What is Lemmatization?
Lemmatization is the process of converting a word into its base, meaningful form (called a *lemma*) using linguistic knowledge.

### Examples:
- running → run  
- ran → run  
- better → good  
- children → child  

💡 Notice:  
"better → good" shows that it understands meaning, not just removing letters.

### How it works:
- Uses **Part of Speech (POS)** (noun, verb, adjective, etc.)
- Uses a **dictionary** to find the correct base form

### Why it matters:
- Without it → words are treated differently ❌  
- With it → same meaning is unified ✅  

🎯 Result:
Better search, smarter chatbots, and more accurate NLP models.

### Key Idea:
“Stemming cuts words. Lemmatization understands them.”

```python
import spacy

# english model
nlp = spacy.load("en_core_web_sm")

# sentence
text = "The children are running and ate better food"

doc = nlp(text)

for token in doc:
    print(token.text, "->", token.lemma_)
```

# Stemming vs Lemmatization (Simple)

| Feature | Stemming | Lemmatization |
|--------|----------|---------------|
| How it works | Removes endings using simple rules | Uses dictionary + POS |
| Result | May not be a real word | Always a real word |
| Accuracy | Low | High |
| Speed | Fast | Slow |
| Problems | Can make mistakes | Needs more computation |

```python
# A list of tricky words, including examples from earlier notes
test_words = ["universe", "university", "alumni", "alumnus", "better", "running", "geese", "was", "are"]

print(f"{'Original Word':<15} | {'Stemming (Porter)':<20} | {'Lemmatization (spaCy)'}")
print("-" * 65)

for word in test_words:
    stem = porter.stem(word)
    lemma = nlp(word)[0].lemma_

    print(f"{word:<15} | {stem:<20} | {lemma}")
```
